## TASK 3 - Rock, Paper, Scissors Game


Objective:
----------
Build a simple Rock-Paper-Scissors game where the player competes with the computer.

This task will test your understanding of:
Input/output
Conditional logic
Random module
Loops
Exception handling

## Instructions:
1.The user should enter their choice: rock, paper, or scissors.

2.The computer should randomly select one of these.

3.Compare both and print who wins.

4.Ask if the player wants to play again- if yes, repeat; if no, exit.

5.Handle invalid inputs gracefully using try-except or condition checks.


```python
import random

choices = ["rock", "paper", "scissors"]

def rock_paper_scissor_game(user_input, system_input):
    if user_input == system_input:
        print("It is a tie")
    elif user_input == "rock" and system_input == "paper":
        print("Sorry! You lose the game")
    elif user_input == "rock" and system_input == "scissors":
        print("Super! You won the game")
    elif user_input == "paper" and system_input == "rock":
        print("Super! You won the game")
    elif user_input == "paper" and system_input == "scissors":
        print("Sorry! You lose the game")
    elif user_input == "scissors" and system_input == "rock":
        print("Sorry! You lose the game")
    elif user_input == "scissors" and system_input == "paper":
        print("Super! You won the game")


while True:
    user_input = input("Enter your choice: rock, paper, or scissors: ").lower()

    if user_input not in choices:
        print("Please enter a valid input.")
        continue

    system_input = random.choice(choices)
    print("System chose:", system_input)

    rock_paper_scissor_game(user_input, system_input)

    user_input_again = input("Do you want to play again? (yes/no): ").lower()

    while user_input_again not in ["yes", "no"]:
        print("Please enter a valid option: yes or no.")
        user_input_again = input("Do you want to play again? (yes/no): ").lower()

    if user_input_again == "no":
        print("Thanks for playing. Exiting the game.")
        break
```

```python
import random

random_words =["python","bridge","coding","engineer","genai","ai","rag","llm"]

system_input = random.choice(random_words)
guess_count = 0
max_guesses = 6 
guessed_letters = []

while guess_count <= max_guesses:
   guess_word = ""
   guess_char = input("Enter one letter: ").lower().strip()
    
   if len(guess_char) != 1 or not guess_char.isalpha():
      print("Please enter exactly one alphabet character.")
   elif guess_char in guessed_letters:
      print("You already guessed that letter.")
   else:
      guessed_letters.append(guess_char)
       
   for char in system_input:
        if char in guessed_letters:
           guess_word = guess_word + char
           print(guess_word)
        else:
           guess_word = guess_word + "_"
           print(guess_word)
   print(f"guessed_letters = {guessed_letters}")
   print(f"wrong_guesses = {max_guesses - guess_count}")

   if "_" not in guess_word:
       print(f"You won the game . the guess word is {guess_word} and system input is {system_input}")
       break
   elif guess_char not in system_input:
      guess_count += 1
      print("Wrong guess!")
   else:
      print("Correct guess!")

else:
    print(f"\nYou lost. The word was: {system_input}")
   
```
