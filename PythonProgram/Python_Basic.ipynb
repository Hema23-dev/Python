## Level3 : Give multiple conditions and number of guess should be less than equal to 7 times.
1. User should enter the number between 1-100. If the entered number is wrong, then print "Invalid number"
 
2. If the User guess number > secret number, then print "your number is greater than secret number"
 
3. If the User guess number < secret number, then print "your number is lesser than secret number"

```python
print("Welcome to number guessing game")
print("Guess the number between 1 to 100")
secret_number = 30
for i in range(1,8):
    user=int(input("enter the guess number:")) 
    if user < 1 or user > 100:
     print("Invalid number. Please enter the number between 1 - 100")
    elif user > secret_number:
     print("your guess number is greater than secret number")
    elif user < secret_number:
     print("your guess number is lesser than secret number")
    elif user == secret_number:
     print("Congratulations! your guess number is correct")
     break
    else:
     print("sorry! your guess is wrong")  
```

## level6: How to handle the invalid characters or values

```python
print("Welcome to number guessing game")
print("Guess the number between 1 to 100")
secret_number = random.randint(1,100)
attempts=0
while True:
    try:
        user=int(input("enter the guess number:")) 
        if user < 1 or user > 100:
         print("Invalid number. Please enter the number between 1 - 100")
         attempts=attempts+1
        elif user > secret_number:
         print("your guess number is greater than secret number")
         attempts=attempts+1  
        elif user < secret_number:
         print("your guess number is lesser than secret number")
         attempts=attempts+1
        elif user == secret_number:
         print("Congratulations! your guess number is correct")
         attempts=attempts+1
         break
    except ValueError:
        print("Invalid character! Not recognized. Please enter the number between 1 - 100")
    

if user == secret_number:
   print(f"Congratulations! your guess number is correct. you have guessed the {secret_number} in {attempts} attempt")
else:       
  print("sorry! you have exceeded all the attempts. Try again") 
```
