## Real-world example:

I want to reuse the same function for Team members and Managers , but they have small differences. 

Both Team members and Managers should be able to log in, but:

1. Team members - should log in, view task,and submit task

2. Managers - should log in , view task and assign task


```python
# defining a function


def login(user_data):
    print(f"the user {user_data['name']} has logged in")

def view_task(user_data):
    print(f"the user {user_data['name']} is viewing the task")

def submit_task(user_data):
    if user_data['role'] == 'teammember':
        print(f"the user {user_data['name']} has submitted the task")
    else:
        print("Managers don't submit the task")

def assign_task(user_data):
    if user_data['role'] == 'manager':
        print(f"the user {user_data['name']} has assigned the task")
    else:
        print("Team member don't assign the task")
    


# data collection

user_data = {"name" : "Hema" , "role" : "manager"}

#functions to be called

login(user_data)
view_task(user_data)
submit_task(user_data)
assign_task(user_data)
```

```python
# for team member

user_data = {"name" : "diya" , "role" : "teammember"}

#functions to be called

login(user_data)
view_task(user_data)
submit_task(user_data)
assign_task(user_data)
```

```python
## In the above code, the user defined function need to be called everytime with data to check the logic, 
## This makes the code no reusable,no inheritance,no structure
```

## With OOPS

class - Blueprint that contains its attributes of particular objects and its all the functions

Object - instance of a class

```python
# CREATE AN CLASS

class employee:
    def __init__(self,name):  # self.name - creates attribute unique to this specific object
        self.name = name
        
    def login(self):
        print(f"the user {self.name} has logged in")

    def view_task(self):
        print(f"the user {self.name} is viewing the task")  

class TeamMember(employee):
    
    def submit_task(self):
        print(f"the user {self.name} has submitted the task")

class Manager(employee):

    def assign_task(self):
        print(f"the user {self.name}  has assigned the task")
  
```

```python
# DEFINE OBJECT

diya = TeamMember("diya")

hema = Manager("hema")
```

```python
hema.login()
```

```python
hema.view_task()
```

```python
hema.assign_task()
```

```python
diya.login()
diya.view_task()
diya.submit_task()
```

## ENCAPSULATION

1. Private : Restricted access from outside class . eg: __atm_pin

2. Public: Can access anywhere outside the class. 

3. protected : accessible to the class  and derived class. eg:_atm_pin

```python
class Secure_user:
    name= None
    atm_pin= None
    
    def __init__(self,name,atm_pin):
        self.name = name
        self._atm_pin = atm_pin
    
    def get_user_pin(self,attempt):
        return attempt == self.atm_pin

    def cash_withdrawal():
        pass

    def deposit_cash():
        pass    
    
```

```python
user_data = Secure_user("hema",2345)   # getting values for the class
```

```python
user_data.get_user_pin(2345) # calling the function of the class with value
```

**But when we give __atm_pin then we will not get the answer itself which is fully secured**
