## Paramters and arguments

```python
def addition(num1,num2,num3): # num1,num2,num3 are called parameters , which is defined in function
    return num1+num2+num3

addition(10,20,30) # passing arguments inside the function ( .. )
```

```python
def get_details(name,place,expected_price):
    print(" Client name :",name)
    print(" Place :" , place)
    print(" expected_price :", expected_price)

# call the function
get_details("Hema","New York", 100000)
```

## Default parameter and Arguments

```python
# to get specific  details from place -> chennai

def get_details_chennai(name,place="chennai",expected_price):
    print(" Client name :",name)
    print(" Place :" , place)
    print(" expected_price :", expected_price)
```

```python
# to get specific  details from place -> chennai

def get_details_chennai(name,expected_price,place="chennai"):
    print(" Client name :",name)
    print(" Place :" , place)
    print(" expected_price :", expected_price)
```

```python
# call the function
get_details_chennai(name="Hema",expected_price=100000) # the here the default place parameter value is taken
```

## Variable argument - it is used when we want one of the parameter to have multiple values

```python
def get_details_chennai(name,place,*expected_price):  # *expected_price - variable argument
    print(" Client name :",name)
    print(" Place :" , place)
    print(" expected_price :", expected_price)
```

```python
# if the expected_price is varying everytime

get_details_chennai("Hema","chennai",100000,900000,200000)
```

## Keyword argument - it is used when we call multiple arugments

```python
def get_details_chennai(name,place,**expected_price):  # **expected_price - keyword argument like price_1,price_2,price_3
    print(" Client name :",name)
    print(" Place :" , place)
    print(" expected_price :", expected_price)
```

```python
get_details_chennai("Hema","chennai",price_1=100000,price_2=900000,price_3=200000)
```

## 5.4 Challenge 2
Find the factors of the number
If the user enters 20 , the output should be "the factors of 20 is 1,2,4,5,10,20

```python
n=int(input("enter the number:"))
res=[]
for i in range(1,n+1):
    if n%i == 0:
       res.append(i)
print(f"the factors of {n} is {res}")
        
    
```

FOR LOOP - for loop is performed when the number of iteration is known
