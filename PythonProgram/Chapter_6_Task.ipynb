## Abstraction


**Abstraction means exposing only necessary behavior and hiding unnecessary implementation details.**

**Abstraction is used when we needs to know what it does and not how it does**

```python
from abc import ABC, abstractmethod

class SecureUser(ABC):          # Abstract base class
    def __init__(self, name, atm_pin):
        self.name = name
        self.__atm_pin = atm_pin   # private-encapsulated data

    def authenticate(self, attempt):
        if attempt == self.__atm_pin:
            print("Proceed with the process")
            return True
        return "Your password is incorrect"

    @abstractmethod
    def account_type(self):        # abstraction: rule without implementation
        pass


class SavingsUser(SecureUser):     # inheritance
    def account_type(self):
        return "Savings Account"


class CurrentUser(SecureUser):     # inheritance
    def account_type(self):
        return "Current Account"
```

## Polymorphism 

**Polymorphism means the same method call behaves differently depending on the object.**

**Here it calls accountype, where when it is called SavingUser then it is Saving Account , where when it is called CurrentUser then it is Current Account**

```python
from abc import ABC, abstractmethod

class SecureUser(ABC):          # Abstract base class
    def __init__(self, name, atm_pin):
        self.name = name
        self.__atm_pin = atm_pin   # private-encapsulated data

    def authenticate(self, attempt):
        if attempt == self.__atm_pin:
            print("Proceed with the process")
            return True
        return "Your password is incorrect"

    @abstractmethod
    def account_type(self):        # abstraction: rule without implementation
        pass


class SavingsUser(SecureUser):     # inheritance
    def account_type(self):
        return "Savings Account"


class CurrentUser(SecureUser):     # inheritance
    def account_type(self):
        return "Current Account"
```

```python
users = [
    SavingsUser("Arun", "1234"),
    CurrentUser("Priya", "5678")
]

for user in users:
    print(user.account_type())     # polymorphism
```
