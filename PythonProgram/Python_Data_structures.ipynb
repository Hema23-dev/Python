## Why Python built-in Data structures
* To collect multiple data in a structure

### Four datastructures in python
1. List
2. Sets
3. Tuples
4. Dictionary - key and value pair


## 1. List
### Properties
* List is one dimensional - 1D
* list is Heterogenous ( accepts different datatypes)
* List accepts duplicate value
* Indexing is possible
* silicing is possible
* Mutable(value can be changed by indexing , silicing and built-in functions)

```python
## defining the empty list
empty_cart = []
```

```python
type(empty_cart)
```

```python
items_to_buy = ["apple","orange","grapes","pomogranate"]
items_to_buy
```

```python
## list accepts duplicate value

items_to_buy = ["apple","orange","grapes","pomogranate","apple"]
items_to_buy
```

```python
items_to_buy = ["apple","orange","grapes","pomogranate",6.7,9,0]
items_to_buy
```

```python
len(items_to_buy)
```

```python
## to retireve 1st element - Indexing

items_to_buy[0]
```

```python
## negative indexing

items_to_buy[-3]
```

```python
#slicing
items_to_buy[0:3]
```

```python
items_to_buy[-6:-1]
```

```python
# changing the value - Mutable
items_to_buy[1] = "watermelon"
items_to_buy
```

## 2. Sets

Properties
-----------
* Set is one dimensional - 1D
* is heterogenous(allow multiple datatype)
* Set does not allow duplicate values
* Indexing and silicing is not possible
* Set is MUTUABLE (by using built-in functions)

```python
# defining empty set

empty_cart2 = set()
empty_cart2
```

```python
type(empty_cart2)
```

```python
items_to_buy2 = {"apple","orange","grapes","pomogranate","apple"}
items_to_buy2
```

```python
type(items_to_buy2)
```

```python
items_to_buy2[0]
```

```python
items_to_buy2[0:4]
```

```python
items_to_buy2[1] = "watermelon"
```

```python
# add an element in set - we need to use built in functions
items_to_buy2.add("banana")
```

```python
items_to_buy2
```

```python
# remove element in set - use function discard to remove the value from set
items_to_buy2.discard("orange")
items_to_buy2
```

## 3. Tuples

Properties
----------
* Tuple is one dimensional 1D
* Heterogenous ( accepts different datatypes)
* allows duplicate
* Indexing and silicing is possible
* Tuple is IMMUTABLE ( WE WILL NOT BE ABLE TO CHANGE THE VALUES OF TUPLE)

```python
# defining empty tuples

empty_cart3 = ()
type(empty_cart3)
```

```python
items_to_buy3 = ("apple","orange","grapes","pomogranate","apple")
items_to_buy3
```

```python
type(items_to_buy3)
```

```python
items_to_buy3[0]
```

```python
items_to_buy3[0:4]
```

```python
items_to_buy3[0] = "bluberry"
```

## 4. Dictionary

It is Key:value pair

Properties
----------
* Dictionary is 2 dimensional 2D
* heterogenous datatype
* In dictionary , don't create duplicate keys in but we can create dupliacte values
* Indexing and silicing is not possible

```python
# defining empty dictionary

empty_cart4 = {}
type(empty_cart4)
```
