## 1. list built-in functions
------------------------------
* append - to add items to the list
  
* copy - shallow copy of the list (copying to new list , by this the orginal list is not modified , if we modify the new list)

* count - returns the number of occurences of the item from the list

* extend - we can add another list or set or tuple or strings of items to the exiting list.

* remove - removes the item from the list

* index - returns the index or position of the item

* insert - to insert or add the item to the list in particular position or index

* pop - removes the last item from the list(last in first out) , if index is mentioned that index item will be removed

* reverse -  reverse the list items from backwards

* sort - sort the list in ascending order

* clear - clears the items in the list

## 2. SET Built-in Functions
----------------------------
* add - adding item to the set ( if already exits , this function do nothing because set does not allow duplicates)
  
* remove - removes the item from the set and if the item is not present in set, it will raise ValueError
 
* copy - shallow copy of the set (copying to new set , by this the orginal set is not modified , if we modify the new set)

* difference - compares two sets and returns the non-common values

* difference_update - compare the two sets and the set1 will be modified by removing the common items between two sets

* discard - removes the itme from the set and if the item not present in set, it will do NOTHING

* intersection - returns the common items between two sets in new set

* intersection_update - compare two sets and keeps the common value in set1 and set2 will be unchanged

* isdisjoint - returns TRUE if they do not share the any one of the item in set otherwise FALSE

* issubset - it is method returns TRUE if the set1 is subset of set2 otherwise FALSE

* issuperset - it is method returns TRUE if the all the values from set1 is present in set2 otherwise FALSE

* pop - pop method in SET will randomly remove the item from the set , but in LIST it pop out last element or with index number it pop out the items in list

* symmetric_difference - returns unique or uncommon value between two sets and returns in new set

* symmetric_difference_update -  set1 will be modified with unique or uncommon value between two sets

* union - set union will combine all the sets unique value in new set. NOTE: it can union LIST values also

* update - method is used to add multiple elements from one or more iterables (like a list, tuple, dictionary, or another set) into an existing set

* clear - removes all elements from a set, leaving it completely empty

## 4.Dictionaries Built-in function - keys:values
-------------------------------------------------
* values - returns the values from the dictionary

* keys - returns the key from the dictionary

* update - adds new key-value pairs or modifies existing ones in a dictionary using elements from another dictionary or an iterable of pairs(list of tuples , keyword argument)

* pop - poping out the element from the dictionary with Key

* popitem - removes the last key:value pair from the dictionary

* copy - shallow copy of the exiting dictionary and further edit to the dicitionary will not affect the exiting dictionary

* fromkeys - used to create a new dictionary from an iterable of keys, assigning a specific common value to all of them

* get - retrieve value of the specified key, if key doesn't exists it returns None or returns the fallback value of if given (such as "not found"...)

* items - return  the key-value pairs of a dictionary as tuples.

* setdefault - retrieve the value of a key if it exists, or insert the key with a specified default value if it does not exist

* clear - clears the dicitonary


```python
stud1_marks={"math":90,"science":80,"tamil":75,"english":85,"social": 60}
print(stud1_marks)
type(stud1_marks)
```

```python
# values
stud1_marks.values()
```

```python
# keys
stud1_marks.keys()
```

```python
# update
stud1_marks.update[("PT",95),("behavior",100)]
stud1_marks
```

```python
# update - updating the dict with list of tuples

new_marks=[("PT",95),("behavior",100)]
stud1_marks.update(new_marks)
stud1_marks
```

```python
# update - updating the dict with keyword argument

stud1_marks.update(artwork=70,handwriting=50)
stud1_marks
```

```python
# if existing key is there , then it will change the value according to new update

stud1_marks.update(PT=88,handwriting=50)
stud1_marks
```

```python
# update - updating with another dictionary

new_marks1={"PT":95,"behavior":100}
stud1_marks.update(new_marks1)
stud1_marks
```

```python
# pop
stud1_marks.pop("handwriting")
stud1_marks
```

```python
# pop 
stud1_marks.pop("handwriting")
stud1_marks
```

```python
# in the above code the KEY doesn't exit, so it returns key value error - to return without any error , we can add fallback as below

stud1_marks.pop("handwriting","Not found")  # if not found , it returns the exiting dict
stud1_marks
```

```python
# popitems

stud1_marks.popitem()
stud1_marks
```

```python
# copy

stud1_marks_backup = stud1_marks.copy()
stud1_marks_backup
```

```python
## fromkeys

keys = ["extra_activity"]
stud1_marks_update = stud1_marks.fromkeys(keys)
stud1_marks_update
```

```python
## get

print(stud1_marks.get("math"))
print(stud1_marks.get("handwriting",0)) # 0 - it gives 0 if key is not found
```

```python
## items - using for loop

for key,value in stud1_marks.items():
    print(f"{key} mark is {value}")
```

```python
## items - to view and retrieve key:value pairs with value update

print(f"stud1_mark dict is {stud1_marks}")
view_stud1_marks = stud1_marks.items()

stud1_marks["extra_activity"] = 70

print(f"viewing dict with items() function - {view_stud1_marks}")
print(f"after change the dict looks like - {stud1_marks}")
```

```python
## setdefault

print(f"stud1_mark dict is {stud1_marks}")

math_score = stud1_marks.setdefault("math",100)

print(f"setting default value in math , stud1_mark dict after change {stud1_marks}") # the math mark will not be changed ,exiting dict value will tabken
```
